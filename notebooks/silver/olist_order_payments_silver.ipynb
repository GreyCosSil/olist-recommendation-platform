```python
volume = '{}/ml_training_dev.olist/raw/files'.format(workspace)
file_name = 'olist_order_payments_dataset.csv'
catalog_name = "ml_training_dev"
schema_name = "silver"
target_table = file_name.split("_dataset.csv")[0]
target_table
```

##Create Catalog and schema

```python
spark.sql("CREATE DATABASE IF NOT EXISTS {}".format(catalog_name))
spark.sql("CREATE SCHEMA IF NOT EXISTS {}".format(schema_name))
```

##Import Libraries

```python
from pyspark.sql.functions import current_date, count,col
```

##Reading Files

```python
raw_df = spark.read.csv(f"{volume}/{file_name}", header=True, inferSchema=True)
raw_df.dropDuplicates()
raw_tranformed = raw_df.withColumn("loadDate", current_date())
```

##Writting table

```python
raw_tranformed.write.mode("overwrite").saveAsTable(f"{catalog_name}.{schema_name}.{target_table}")
print("table successfully created!")
```
