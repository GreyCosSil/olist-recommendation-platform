```python
def records_df(df1, df2, result) -> bool:
    df1_size= df1.count()
    df2_size= df2.count()
    result_size= result.count()

    bigger_df = df2_size if df1_size < df2_size else df1_size

    if result_size > bigger_df:
        return  True
    else:
        return False
```
